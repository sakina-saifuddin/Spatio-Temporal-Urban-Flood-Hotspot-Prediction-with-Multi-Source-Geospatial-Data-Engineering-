```python
yearly_locations = {}

for year in [2015, 2016, 2017, 2019]:
    path = f"../data/raw/city_of_houston/floodingdrainage{year}.csv"
    yearly_df = pd.read_csv(path)

    aggregated_df = (
        yearly_df.groupby(["lat", "lon"])
        .size()
        .reset_index(name="report_count")
    )

    aggregated_df["year"] = year
    yearly_locations[year] = aggregated_df

    print(f"{year}: {len(aggregated_df)} unique locations")
```

```python
for year in [2015, 2016, 2017, 2019]:
    df_year = yearly_locations[year]

    print(f"Year: {year}")
    print("Unique locations:", len(df_year))
    print("Total records represented:", df_year["report_count"].sum())
    print()
    
```

```python
from pathlib import Path

data_folder = Path("../data")

for file in data_folder.rglob("*"):
    if file.suffix.lower() in [".shp", ".geojson", ".gpkg"]:
        print(file)
```

```python
import geopandas as gpd

houston_boundary = gpd.read_file("../data/boundaries/houston_city_limits.geojson")

houston_boundary.head()
```

```python
print("Coordinate Reference System:", houston_boundary.crs)

print("\nGeometry types:")
print(houston_boundary.geometry.geom_type.value_counts())

print("\nService area types:")
print(houston_boundary["SERVICE_TY"].value_counts(dropna=False))
```

```python
houston_city_geometry = houston_boundary.geometry.union_all()

print("Combined geometry type:", houston_city_geometry.geom_type)
print("Is geometry valid:", houston_city_geometry.is_valid)
print("Is geometry empty:", houston_city_geometry.is_empty)
```

```python
flood_points_2015 = gpd.GeoDataFrame(
    yearly_locations[2015].copy(),
    geometry=gpd.points_from_xy(
        yearly_locations[2015]["lon"],
        yearly_locations[2015]["lat"]
    ),
    crs="EPSG:4326"
)

flood_points_2015.head()
```

```python
inside_houston = flood_points_2015.geometry.intersects(houston_city_geometry)

print("Locations inside Houston:", inside_houston.sum())
print("Locations outside Houston:", (~inside_houston).sum())
print("Total locations:", len(flood_points_2015))
```

```python
flood_points_2015["inside_houston"] = inside_houston

print(flood_points_2015["inside_houston"].value_counts())
```

```python
outside_records = flood_points_2015.loc[
    ~flood_points_2015["inside_houston"],
    "report_count"
].sum()

inside_records = flood_points_2015.loc[
    flood_points_2015["inside_houston"],
    "report_count"
].sum()

print("Original records inside Houston:", inside_records)
print("Original records outside Houston:", outside_records)
print("Total original records:", inside_records + outside_records)
```

```python
yearly_flood_points = {2015: flood_points_2015}

for year in [2016, 2017, 2019]:
    df_year = yearly_locations[year]

    flood_points = gpd.GeoDataFrame(
        df_year.copy(),
        geometry=gpd.points_from_xy(
            df_year["lon"],
            df_year["lat"]
        ),
        crs="EPSG:4326"
    )

    flood_points["inside_houston"] = flood_points.geometry.intersects(
        houston_city_geometry
    )

    yearly_flood_points[year] = flood_points

    print(f"Year: {year}")
    print("Inside locations:", flood_points["inside_houston"].sum())
    print("Outside locations:", (~flood_points["inside_houston"]).sum())
    print()
```

```python
for year in [2015, 2016, 2017, 2019]:
    flood_points = yearly_flood_points[year]

    inside_records = flood_points.loc[
        flood_points["inside_houston"],
        "report_count"
    ].sum()

    outside_records = flood_points.loc[
        ~flood_points["inside_houston"],
        "report_count"
    ].sum()

    print(f"Year: {year}")
    print("Original records inside Houston:", inside_records)
    print("Original records outside Houston:", outside_records)
    print("Total original records:", inside_records + outside_records)
    print()
```

```python
cleaned_flood_points = {}

for year in [2015, 2016, 2017, 2019]:
    flood_points = yearly_flood_points[year]

    cleaned_df = flood_points[
        flood_points["inside_houston"]
    ].copy()

    cleaned_flood_points[year] = cleaned_df

    print(f"Year: {year}")
    print("Remaining locations:", len(cleaned_df))
    print("Original records represented:", cleaned_df["report_count"].sum())
    print()
```

```python
flood_points_2015_meters = cleaned_flood_points[2015].to_crs("EPSG:32615")

print("Original CRS:", cleaned_flood_points[2015].crs)
print("Projected CRS:", flood_points_2015_meters.crs)

flood_points_2015_meters.head()
```

```python
spatial_index = flood_points_2015_meters.sindex

nearby_pairs = spatial_index.query(
    flood_points_2015_meters.geometry,
    predicate="dwithin",
    distance=50
)

print("Nearby pairs found:", nearby_pairs.shape[1])
print("First five pairs:")
print(nearby_pairs[:, :5])
```

```python
first_indices = nearby_pairs[0]
second_indices = nearby_pairs[1]

unique_pair_mask = first_indices < second_indices

unique_nearby_pairs = nearby_pairs[:, unique_pair_mask]

print("Total spatial matches:", nearby_pairs.shape[1])
print("Distinct nearby-location pairs:", unique_nearby_pairs.shape[1])
print("First five distinct pairs:")
print(unique_nearby_pairs[:, :5])
```

```python
first_points = flood_points_2015_meters.geometry.iloc[
    unique_nearby_pairs[0]
].reset_index(drop=True)

second_points = flood_points_2015_meters.geometry.iloc[
    unique_nearby_pairs[1]
].reset_index(drop=True)

pair_distances = first_points.distance(second_points)

print("Minimum distance:", pair_distances.min())
print("Maximum distance:", pair_distances.max())
print("Average distance:", pair_distances.mean())
print("Median distance:", pair_distances.median())
```

```python
import numpy as np

locations_with_neighbors = np.unique(unique_nearby_pairs)

print("Total 2015 locations:", len(flood_points_2015_meters))
print("Locations with a neighbor within 50m:", len(locations_with_neighbors))
print(
    "Locations without a neighbor within 50m:",
    len(flood_points_2015_meters) - len(locations_with_neighbors)
)
```

```python
for year in [2016, 2017, 2019]:
    flood_points_meters = cleaned_flood_points[year].to_crs("EPSG:32615")

    nearby_pairs_year = flood_points_meters.sindex.query(
        flood_points_meters.geometry,
        predicate="dwithin",
        distance=50
    )

    unique_pairs_year = nearby_pairs_year[
        :,
        nearby_pairs_year[0] < nearby_pairs_year[1]
    ]

    locations_with_neighbors_year = np.unique(unique_pairs_year)

    print(f"Year: {year}")
    print("Total locations:", len(flood_points_meters))
    print("Distinct nearby pairs:", unique_pairs_year.shape[1])
    print("Locations with nearby neighbors:", len(locations_with_neighbors_year))
    print()
```

```python
export_columns = ["lat", "lon", "report_count", "year"]

export_datasets = {}

for year in [2015, 2016, 2017, 2019]:
    export_df = cleaned_flood_points[year][export_columns].copy()

    export_datasets[year] = export_df

    print(f"Year: {year}")
    print("Rows:", len(export_df))
    print("Columns:", export_df.columns.tolist())
    print()
```

```python
from pathlib import Path

output_directory = Path("../data/processed")
output_directory.mkdir(parents=True, exist_ok=True)

for year in [2015, 2016, 2017, 2019]:
    output_path = output_directory / f"houston_311_cleaned_{year}.csv"

    export_datasets[year].to_csv(output_path, index=False)

    print(f"Saved: {output_path}")
```

```python
for year in [2015, 2016, 2017, 2019]:
    file_path = output_directory / f"houston_311_cleaned_{year}.csv"

    saved_df = pd.read_csv(file_path)
    original_df = export_datasets[year]

    print(f"Year: {year}")
    print("Saved rows:", len(saved_df))
    print("Columns match:", saved_df.columns.tolist() == original_df.columns.tolist())
    print("Data matches:", saved_df.equals(original_df))
    print("Record count total:", saved_df["report_count"].sum())
    print()
```

```python
for year in [2015, 2016, 2017, 2019]:
    file_path = output_directory / f"houston_311_cleaned_{year}.csv"

    saved_df = pd.read_csv(file_path)
    original_df = export_datasets[year]

    print(f"Year: {year}")
    print("Original index:", original_df.index[:5].tolist())
    print("Saved index:", saved_df.index[:5].tolist())

    print(
        "Data matches after resetting index:",
        saved_df.equals(original_df.reset_index(drop=True))
    )
    print()
```

```python
cleaning_summary = []

for year in [2015, 2016, 2017, 2019]:
    original_df = pd.read_csv(
        f"../data/raw/city_of_houston/floodingdrainage{year}.csv"
    )

    cleaned_df = export_datasets[year]

    cleaning_summary.append({
        "year": year,
        "original_rows": len(original_df),
        "unique_locations_before_filter": len(yearly_locations[year]),
        "locations_after_filter": len(cleaned_df),
        "locations_removed": (
            len(yearly_locations[year]) - len(cleaned_df)
        ),
        "original_rows_represented_after_filter": cleaned_df["report_count"].sum(),
        "original_rows_excluded": (
            len(original_df) - cleaned_df["report_count"].sum()
        )
    })

cleaning_summary_df = pd.DataFrame(cleaning_summary)

display(cleaning_summary_df)
```

```python
summary_path = output_directory / "houston_311_cleaning_summary.csv"

cleaning_summary_df.to_csv(summary_path, index=False)

print("Saved cleaning summary:", summary_path)
```
